## 2. Handling Missing Values

The `CouponCode` column contains missing values. Since a missing coupon code indicates that no coupon was provided or used, these records should not be deleted.

The missing values will be replaced with `NO_COUPON` to make the information explicit while preserving all order records.

```python
print("Missing CouponCode values before cleaning:",
      df["CouponCode"].isna().sum())
```

```python
df["CouponCode"] = df["CouponCode"].fillna("NO_COUPON")
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
print("Duplicate OrderIDs after cleaning:",
      df["OrderID"].duplicated().sum())
```

```python
date_check = pd.to_datetime(df["Date"], errors="coerce")

print("Invalid dates after cleaning:", date_check.isna().sum())
```

```python
numeric_columns = ["Quantity", "UnitPrice", "ItemsInCart", "TotalPrice"]

for column in numeric_columns:
    invalid = pd.to_numeric(df[column], errors="coerce").isna().sum()
    print(column, "→ invalid values:", invalid)
```

```python
print("Total missing values in cleaned dataset:",
      df.isnull().sum().sum())
```

```python
# STEP 26A - Export Final Cleaned Dataset

cleaned_file = "cleaned_dataset.xlsx"

df.to_excel(cleaned_file, index=False)

print("Final cleaned Excel file created successfully!")
print("File name:", cleaned_file)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```
